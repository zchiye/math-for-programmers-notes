```python
# 练习7.2
width = 1920
height = 1080

def transform(x, y, minx=-10, maxx=10, miny=-10, maxy=10, width=width, height=height):
    """将(x,y)从(minx,maxx,miny,maxy)映射到(0,width,0,height)"""
    x = (x - minx) / (maxx - minx) * width
    y = (maxy - y) / (maxy - miny) * height
    return int(x), int(y)

print(transform(0, 0))
print(transform(-10, -10))
print(transform(10, 10))
print(transform(10, -10))

import random

print(transform(random.uniform(-10, 10), random.uniform(-10, 10)))
```
