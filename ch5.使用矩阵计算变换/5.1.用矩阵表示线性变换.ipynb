```python
# 矩阵与向量乘法
from vectors import *

def matrix_multi_vector(matrix, vector):
    return linear_combination(vector, *zip(*matrix))

matrix_B = (
    (0,2,1),
    (0,1,0),
    (1,0,-1)
)
v = (3,-2,5)

print(matrix_multi_vector(matrix_B, v))
```

```python
# 矩阵与向量乘法  使用点积实现
def matrix_multi_vector_dot(matrix, vector):
    return tuple(dot(row, vector) for row in matrix)

print(matrix_multi_vector_dot(matrix_B, v))
```

```python
# 矩阵乘法

def matrix_multiply(a,b):
    return tuple(
        tuple(dot(row,col) for col in zip(*b))
        for row in a
    )

A = (
    (1,1,0),
    (1,0,1),
    (1,-1,1)
)
C = (
    (0,2,1),
    (1,0,1),
    (1,0,-1)
)
C_reverse = (
    (0,0,1),
    (2,1,0),
    (1,0,-1)
)

print(matrix_multiply(A,C))
print(matrix_multiply(A,C_reverse))
```

```python
# 用矩阵变换表示三维动画
from teapot import load_triangles
from draw_model import draw_model
from math import sin,cos

def get_rotation_matrix(t):      # 为任意表示时间的数字参数生成一个新的变换矩阵
    seconds = t/1000      # 将时间单位转换为秒，使变换不会发生得太快
    return (
        (cos(seconds),0,-sin(seconds)),
        (0,1,0),
        (sin(seconds),0,cos(seconds))
    )

draw_model(load_triangles(),
           get_matrix=get_rotation_matrix)      # 将函数作为关键字参数传递给函数draw_model
```

```python
# 练习5.1：实现函数infer_matrix(n, transformation)，接收一个维度参数（比如2或3）和一个线性向量变换的函数参数，返回[插图]方阵（一个[插图]元组的[插图]元组的数字集，表示线性变换的矩阵）​。
def infer_matrix(n, transformation):
    def standard_basis_vector(i):
        return tuple(1 if i==j else 0 for j in range(1,n+1))     # 创建第i个标准基向量表示一个元组，在第i个坐标中包含1，在所有其他坐标中包含0
    standard_basis = [standard_basis_vector(i) for i in range(1,n+1)]     # 创建标准基表示n个向量的列表
    cols = [transformation(v) for v  in standard_basis]     # 将矩阵的列定义为对标准基向量进行相应线性变换的结果
    return tuple(zip(*cols))     # 按照惯例，将矩阵重构为行元组，而不是列的列表

from transforms import *
from math import pi

print(infer_matrix(3, rotate_z_by(pi/2)))    # 旋转90度
```

```python
def infer_rotation_matrix(t):
    return infer_matrix(3, lambda v: rotate_z_by(t/1000)(v))    # 旋转角度与时间成正比

draw_model(load_triangles(),
           get_matrix=infer_rotation_matrix)   # 传递函数infer_rotation_matrix作为关键字参数
```

```python
# 练习5.3
import random

def random_matrix(n):
    return tuple(
        tuple(random.randint(-100, 101) for _ in range(n))
        for _ in range(n)
    )

for _ in range(5):
    print(random_matrix(3))
```
