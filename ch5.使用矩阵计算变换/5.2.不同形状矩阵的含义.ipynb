```python
# 用矩阵乘法表示矩阵×向量

from vectors import *

def matrix_multiply(a,b):
    return tuple(
        tuple(dot(row,col) for col in zip(*b))
        for row in a
    )

matrix_B = (
    (0,2,1),
    (0,1,0),
    (1,0,-1)
)
matrix_V = (
    (3,),
    (-2,),
    (5,)
)

print(matrix_multiply(matrix_B, matrix_V))
```

```python
# 会被自动转换为1层的元组  ((3,-2,5)) -> (3,-2,5)
# matrix_row = (
#     (3, -2, 5)
# )

# 正确写法
matrix_row = (
    (3, -2, 5),
)

print(matrix_multiply(matrix_row, matrix_B))
```

```python
# 非方阵的矩阵乘法

matrix_A = (
    (1,-2,0),
    (-1,-2,2)
)
matrix_B = (
    (2,0,-1,2),
    (0,-2,2,-2),
    (-1,-1,2,1)
)

print(matrix_multiply(matrix_A, matrix_B))
```

```python
# 将投影函数用矩阵表示

matrix_project_2d_xy = (
    (1,0,0),
    (0,1,0)
)

vector_xyz = (
    (3,),
    (5,),
    (7,)
)

print(matrix_multiply(matrix_project_2d_xy, vector_xyz))
```

```python
# 练习5.18：实现一个函数，将列向量转换成行向量，或者将行向量转换成列向量。像这样把一个矩阵翻转过来叫作转置，而得到的矩阵叫作原矩阵的转置矩阵
def transpose(m):
    return tuple(zip(*m))

print(transpose(matrix_1))
print(transpose(matrix_2))
```
